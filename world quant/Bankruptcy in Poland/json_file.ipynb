## Import

Now that we have everything set up the way we need it to be, let's combine all these steps into a single function that 
will decompress the file, load it into a DataFrame, and return it to us as something we can use.

```python
## Create a wrangle function that takes the name of a compressed file as input and returns a tidy DataFrame. 
## After you confirm that your function is working as intended.


def wrangle(filename):
    # Open compressed file, load into dict
    with gzip.open(filename, "r") as f:
        data = json.load(f)

    # Turn dict into DataFrame
    df = pd.DataFrame().from_dict(data["data"]).set_index("company_id")
    return df
```

```python
df = wrangle("data/poland-bankruptcy-data-2009.json.gz")
print(df.shape)
df.head()
```
