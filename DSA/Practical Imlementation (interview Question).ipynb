```python
def binarysearch(arr, i, j, x):
    while i <= j:
        mid = i + (j-i) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            return binarysearch(arr, mid+1, j, x)
        else:
            return binarysearch(arr, i, mid-1, x)
    return -1

arr = [20, -30, 10, 5, 7, 0, 29, float('inf'), float('inf'), float('inf') ]
arr = sorted([20, -30, 10, 5, 7, 0, 29, float('inf'), float('inf'), float('inf') ])

x = float('inf')
i = 0
j = len(arr) -1

result = binarysearch(arr, i, j, x)
print("Searching element is present at the index", result)
        
```

```python
## Implementation of Binary Search using Recursion
## Function Definition
## Time Complexity : O(logn)
def binarySearch(arr, i, j, x):
    while i <= j:
        mid = i + (j-i) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            ## Recursion -> calling the same function again with differnt set of parameters
            return binarySearch(arr, mid+1, j, x)
        else:
            return binarySearch(arr, i, mid-1, x)
    ## Searching element is not present in the array
    return -1



## Driver Code
## Sorted array

arr = sorted([20, -30, 10, 5, 7, 0, 29, float('inf'), float('inf'), float('inf')]) 
x = float('inf')
i = 0
j = len(arr) - 1

## Function call
result = binarySearch(arr, i, j, x)
print("Searching element is present at the index", result)
```

```python
arr = [20, "-30,", 10, 5, 7, 0, 29, float('inf'), float('inf'), float('inf')]

def find_inf(array):
    for i in range(len(array)):
        if array[i] == float('inf'):
            return i

# Call the function
print(find_inf(arr))  # Output: 7
```
