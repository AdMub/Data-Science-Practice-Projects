### **Simple way of querying**

```python
## Simple way of querying
empinfo.find_one()
```

```python
## Select * from employeeinformation on sql
empinfo.find()
```

```python
## Select * from employeeinformation
for record in empinfo.find():
    print(record)
```

```python
## Select * from employeeinformation
for record in empinfo.find({}):
    print(record)
```

```python
## Query the json documents based on equality conditions
# Select * from employeeinformation where firstname=AdMub

for record in empinfo.find({'firstname':'AdMub'}):
    print(record)
```

```python
## Query documents using query operators($in,$lt,$gt)
# select * emplyeeinformation where your qualification isin BTech, phd, master

for record in empinfo.find({'qualification':{'$in':["BTech",'phd','master']}}):
    print(record)
```

```python
## And and Query operators 
# "$lt" means less than
# select * emplyeeinformation where your qualification isin BTech and age 32 

for record in empinfo.find({'qualification':'master','age':{'$lt':35}}):
    print(record)
```

```python
for record in empinfo.find({'qualification':'BTech','age':{'$lt':35}}):
    print(record)
```

```python
## OR operators
for record in empinfo.find({'$or':[{'firstname':'AdMub'},{'qualification':'phd'}]}):
    print(record)
```

```python
## OR operators
for record in empinfo.find({'$and':[{'firstname':'AdMub'},{'Skills':'Data Science'}]}):
    print(record)
```
