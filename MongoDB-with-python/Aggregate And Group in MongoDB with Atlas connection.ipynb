```python
# Access database  
mydatabase = client['Students']  
    
# Access collection of the database  
collection=mydatabase['studentscores']  
data = [ 
    {"user":"AdMub", "subject":"Database", "score":80}, 
    {"user":"AdMal",  "subject":"JavaScript", "score":90}, 
    {"user":"Adroq",  "title":"Database", "score":85}, 
    {"user":"Hamid",  "title":"JavaScript", "score":75}, 
    {"user":"AdMal",  "title":"Data Science", "score":60},
    {"user":"AdMub",  "title":"Data Science", "score":95}] 
  
collection.insert_many(data)
```

```python
### Find AdMal And AdMub Total Subjects
#"_id" unique Identity (it's not compulsory to use"_id") because it is unique value in mongodb compass
#"$user" field name


agg_result= collection.aggregate( 
    [{ 
    "$group" :  
        {"_id" : "$user",  
         "Total Subject" : {"$sum" : 1} 
         }} 
    ]) 
```

```python
for i in agg_result: 
    print(i)
```

```python
### Calculating the total score based on user
agg_result= collection.aggregate( 
    [{ 
    "$group" :  
        {"_id" : "$user",  
         "Total Marks" : {"$sum" :"$score"} 
         }} 
    ]) 
for i in agg_result: 
    print(i)
```

```python
### Calculating the average score based on user
agg_result=collection.aggregate([
   {
      "$group": {
         "_id": '$user',
         "StudentScoreAverage": {
            "$avg": "$score"
         }
      }
   }
])
for i in agg_result: 
    print(i)
```

```python
### Calculating the average quantity And Average Price
agg_result=mycollection.aggregate([
   {
      "$group": {
         "_id": '$item',
         "avgAmount": {"$avg": {"$multiply": [ "$price", "$quantity" ]}},
          "avgQuantity": { "$avg": "$quantity" }
      }
   }
])
for i in agg_result: 
    print(i)
```
