#**Stemming**

```python
# Download the stopwords resource
nltk.download('stopwords')
```

```python
# import necessary Libraries
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords # Remove some common words
```

```python
sent = nltk.sent_tokenize(paragraph)
stemmer = PorterStemmer()  # Initialize PorterStemmer
```

```python
stopwords.words('english') # English stopwords
```

```python
stopwords.words('arabic') # Arabic stopwords
```

```python
# Stemming
for i in range(len(sent)):
  words = nltk.word_tokenize(sent[i])
  words = [stemmer.stem(word) for word in words if word not in set(stopwords.words("english"))]
  sent[i] = ' '.join(words)
```

```python
print(sent)
```

###**Problem of Stemming:**
Produced intermediate representation of the word may not have any meaning.

Example: intelligen, fina

#**Lemmatization**

```python
# import necessary Libraries
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
```

```python
sent = nltk.sent_tokenize(paragraph)
lemmatizer = WordNetLemmatizer()     # WordNet Lemmatizer
```

```python
# Lemmatizion
for i in range(len(sent)):
  words = nltk.word_tokenize(sent[i])
  words = [lemmatizer.lemmatize(word) for word in words if word not in set(stopwords.words("english"))]
  sent[i] = ' '.join(words)
```

```python
# import necesary libraries
from gensim.models import Word2Vec
from nltk.corpus import stopwords
import re
```

```python
# Preprocessing the data
text = re.sub(r'\[[0-9]*\]', ' ', paragraph)
text = re.sub(r'\s+', ' ', text) # rmoving unnecessary spaces
text = text.lower()
text = re.sub(r'\d', ' ', text)
text = re.sub(r'\s+', ' ', text)
```

```python
print(text)
```

```python
# Preparing  the dataset
sent = nltk.sent_tokenize(text)
```

```python
sent = [nltk.word_tokenize(sentence) for sentence in sent]
print(sent)
```

```python
for i in range(len(sent)):
  sent[i] = [word for word in sent[i] if word not in stopwords.words('english')]
```

```python
# Training the Word2Vec model
model = Word2Vec(sent, min_count=1)
```

```python
#words = model.wv.vocab

# In Gensim 4.0.0 and later, .wv.vocab was removed.
# Use .wv.index_to_key to get a list of words in the vocabulary.
words = model.wv.index_to_key
print(words)
```

```python
# Finding Word Vectors
vector = model.wv['war']
```
